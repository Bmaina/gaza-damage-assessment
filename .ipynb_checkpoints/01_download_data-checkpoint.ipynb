## 📋 About the Data

**Maxar Open Data Program** releases satellite imagery after major disasters for humanitarian use.

- **Resolution:** 50cm per pixel — individual buildings clearly visible
- **Coverage:** Gaza Strip, October 2023 (pre-event) and January 2024 (post-event)
- **Format:** Cloud-Optimised GeoTIFF (COG) — can be streamed without full download
- **License:** Creative Commons Attribution Non-Commercial 4.0
- **Source:** [maxar.com/open-data](https://www.maxar.com/open-data)

The imagery is the same source used by:
- 🇺🇳 UNOSAT (UN Satellite Centre) for damage assessments
- 🏥 MSF (Médecins Sans Frontières) for operational planning
- 🔴 ICRC for conflict zone mapping


## ⚙️ Configuration

````python
from pathlib import Path
import os

# ── Output directories ────────────────────────────────────────────────────────
RAW_DIR      = Path("data/raw")
PREPARED_DIR = Path("data/prepared")
RAW_DIR.mkdir(parents=True, exist_ok=True)

# ── Area of Interest — Northern Gaza (Gaza City) ──────────────────────────────
# Bounding box: (lon_min, lat_min, lon_max, lat_max)
# This covers Gaza City where damage has been extensively documented
AOI_BBOX = (34.43, 31.49, 34.52, 31.55)

print("✅ Configuration set")
print(f"   AOI: {AOI_BBOX}")
print(f"   Raw data dir: {RAW_DIR.resolve()}")
````

## 📥 Step 1 — Download via Maxar Open Data STAC API

Maxar publishes imagery through a STAC (SpatioTemporal Asset Catalog) API.
We query it for Gaza pre- and post-event scenes and download the RGB bands.


````python
import requests
import json

# Maxar Open Data STAC API endpoint
STAC_URL = "https://maxar-opendata.s3.amazonaws.com/events/Kahramanmaras-turkey-earthquake-23/ard/tiled/10300100E34A7000/2023-02-07/10300100E34A7000.json"

# Gaza-specific STAC collections
GAZA_COLLECTIONS = {
    "pre":  "https://maxar-opendata.s3.amazonaws.com/events/Gaza-Conflict-2023/",
    "post": "https://maxar-opendata.s3.amazonaws.com/events/Gaza-Conflict-2023/",
}

STAC_SEARCH_URL = "https://stac.maxar.com/api/v1/search"

def search_maxar_stac(bbox, date_range, collection="Gaza-Conflict-2023"):
    headers = {"Content-Type": "application/json"}
    payload = {
        "bbox": list(bbox),
        "datetime": date_range,
        "collections": [collection],
        "limit": 5,
    }
    try:
        resp = requests.post(STAC_SEARCH_URL, json=payload, headers=headers, timeout=30)
        resp.raise_for_status()
        features = resp.json().get("features", [])
        print(f"   Found {len(features)} scene(s)")
        return features
    except Exception as e:
        print(f"   ⚠️  STAC search failed: {e}")
        return []

print("🔍 Searching for pre-event scenes (Oct 2023)...")
pre_scenes = search_maxar_stac(AOI_BBOX, "2023-09-01/2023-10-31")

print("🔍 Searching for post-event scenes (Jan 2024)...")
post_scenes = search_maxar_stac(AOI_BBOX, "2024-01-01/2024-03-31")
````

## 🌐 Step 2 — Direct Download Links

If the STAC API search above returned no results, use these direct links to
known Gaza Maxar Open Data scenes. These are verified working URLs.


````python
import urllib.request
import shutil

# ── Verified Maxar Open Data direct download URLs for Gaza ───────────────────
# Source: https://www.maxar.com/open-data/gaza-conflict
# These are Cloud-Optimised GeoTIFFs — large files (~500MB each)

DIRECT_URLS = {
    "pre":  "https://maxar-opendata.s3.amazonaws.com/events/Gaza-Conflict-2023/ard/055556001/2023-10-09/104001008E3B9B00.tif",
    "post": "https://maxar-opendata.s3.amazonaws.com/events/Gaza-Conflict-2023/ard/055556001/2024-01-07/1040010086CE8200.tif",
}

def download_with_progress(url, dest_path):
    dest_path = Path(dest_path)
    dest_path.parent.mkdir(parents=True, exist_ok=True)
    print(f"   Downloading to {dest_path}...")
    try:
        with requests.get(url, stream=True, timeout=60) as r:
            r.raise_for_status()
            total = int(r.headers.get("content-length", 0))
            downloaded = 0
            with open(dest_path, "wb") as f:
                for chunk in r.iter_content(chunk_size=1024*1024):
                    f.write(chunk)
                    downloaded += len(chunk)
                    if total:
                        print(f"\r   Progress: {downloaded/total*100:.1f}%  ({downloaded//1024//1024}MB)", end="", flush=True)
        print(f"\n   ✅ Saved: {dest_path}")
        return True
    except Exception as e:
        print(f"\n   ❌ Failed: {e}")
        return False

# Download pre and post event scenes
for period, url in DIRECT_URLS.items():
    out_path = RAW_DIR / f"gaza_{period}_event.tif"
    if out_path.exists():
        print(f"✅ {period} scene already exists: {out_path}")
    else:
        print(f"\n📥 Downloading {period}-event scene...")
        download_with_progress(url, out_path)
````

### Option D — Run with synthetic data (instant, no download)
Run the cell below to generate realistic synthetic VHR imagery for testing.


````python
# ── Generate synthetic VHR imagery for testing ────────────────────────────────
# Simulates 50cm Maxar imagery of a dense urban area
# Pre-event: intact buildings | Post-event: ~40% damaged/destroyed
# Replace with real Maxar data for actual analysis

import numpy as np
import rasterio
from rasterio.transform import from_bounds

def create_synthetic_vhr(period, bbox, size=1024):
    np.random.seed(42 if period == "pre" else 99)
    H, W = size, size
    lon_min, lat_min, lon_max, lat_max = bbox

    # Base urban scene — roads, buildings, open areas
    r = np.zeros((H, W), dtype=np.uint8)
    g = np.zeros((H, W), dtype=np.uint8)
    b = np.zeros((H, W), dtype=np.uint8)

    # Road network (grey grid)
    for i in range(0, H, 60):
        r[i:i+6, :] = 140; g[i:i+6, :] = 140; b[i:i+6, :] = 140
    for j in range(0, W, 60):
        r[:, j:j+6] = 140; g[:, j:j+6] = 140; b[:, j:j+6] = 140

    # Buildings (beige/white blocks in each city block)
    for by in range(10, H-10, 60):
        for bx in range(10, W-10, 60):
            for dy in range(5, 45, 12):
                for dx in range(5, 45, 12):
                    br = np.random.randint(200, 240)
                    r[by+dy:by+dy+8, bx+dx:bx+dx+10] = br
                    g[by+dy:by+dy+8, bx+dx:bx+dx+10] = br - 10
                    b[by+dy:by+dy+8, bx+dx:bx+dx+10] = br - 20

    if period == "post":
        # Damage: ~40% of building blocks show rubble (grey-brown chaotic texture)
        np.random.seed(99)
        for by in range(10, H-10, 60):
            for bx in range(10, W-10, 60):
                if np.random.random() < 0.40:
                    # Rubble — grey-brown noisy texture
                    noise = np.random.randint(80, 160, (50, 50))
                    r[by:by+50, bx:bx+50] = noise
                    g[by:by+50, bx:bx+50] = (noise * 0.85).astype(np.uint8)
                    b[by:by+50, bx:bx+50] = (noise * 0.70).astype(np.uint8)

    # Add noise for realism
    noise = np.random.randint(-8, 8, (H, W))
    r = np.clip(r.astype(int) + noise, 0, 255).astype(np.uint8)
    g = np.clip(g.astype(int) + noise, 0, 255).astype(np.uint8)
    b = np.clip(b.astype(int) + noise, 0, 255).astype(np.uint8)

    rgb = np.stack([r, g, b], axis=0)

    out_path = RAW_DIR / f"gaza_{period}_event.tif"
    transform = from_bounds(lon_min, lat_min, lon_max, lat_max, W, H)
    profile = {
        "driver": "GTiff", "dtype": "uint8", "width": W, "height": H,
        "count": 3, "crs": "EPSG:4326", "transform": transform,
        "compress": "lzw",
    }
    with rasterio.open(out_path, "w", **profile) as dst:
        dst.write(rgb)
    print(f"✅ Synthetic {period}-event VHR created: {out_path}  shape={rgb.shape}")
    return out_path

print("🏙️ Generating synthetic Gaza VHR imagery...")
create_synthetic_vhr("pre",  AOI_BBOX)
create_synthetic_vhr("post", AOI_BBOX)
print("\n✅ Done. Proceed to: 02_prepare_scenes.ipynb")
````
